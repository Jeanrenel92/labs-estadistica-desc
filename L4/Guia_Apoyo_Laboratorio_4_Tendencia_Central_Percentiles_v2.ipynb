## 1. Crear un conjunto de datos de ejemplo


````python
import pandas as pd

datos = {
    'Estudiante': ['Ana', 'Bruno', 'Carla', 'Diego', 'Elena', 'Felipe',
                   'Gabriela', 'Hugo', 'Isabel', 'Javier'],
    'Carrera': ['Informática', 'Electricidad', 'Informática', 'Electricidad',
                'Informática', 'Electricidad', 'Informática', 'Electricidad',
                'Informática', 'Electricidad'],
    'Jornada': ['Diurna', 'Diurna', 'Vespertina', 'Vespertina', 'Diurna',
                'Diurna', 'Vespertina', 'Vespertina', 'Diurna', 'Vespertina'],
    'Edad': [19, 22, 20, 24, 21, 23, 20, 25, 22, 24],
    'Puntaje': [62, 75, 68, 80, 72, 75, 90, 65, 75, 78]
}

df = pd.DataFrame(datos)
df
````

## 2. Explorar los datos

Antes de calcular estadísticas, conviene conocer la estructura del DataFrame.

- `head()` muestra las primeras filas.
- `shape` entrega `(número de filas, número de columnas)`.
- `columns` muestra los nombres de las variables.
- `info()` muestra tipos de datos y valores no nulos.


````python
display(df.head())
print("Dimensiones:", df.shape)
print("Columnas:", list(df.columns))
df.info()
````

## 3. Resumen estadístico rápido con `describe()`

Antes de calcular cada medida por separado, Pandas permite obtener varias estadísticas básicas de una variable numérica con un solo comando:

```python
df['Puntaje'].describe()
```

Este comando entrega:

| Resultado | Significado |
|---|---|
| `count` | cantidad de observaciones |
| `mean` | media aritmética |
| `std` | desviación estándar |
| `min` | valor mínimo |
| `25%` | primer cuartil o percentil 25 |
| `50%` | mediana o percentil 50 |
| `75%` | tercer cuartil o percentil 75 |
| `max` | valor máximo |

> **Importante:** `describe()` no entrega la moda. Para obtenerla utilizaremos `.mode()`.


````python
df['Puntaje'].describe()
````

### ✏️ ¿Cómo leer el resultado?

Por ejemplo, si en la fila `mean` aparece `74.0`, significa que el **puntaje promedio** es 74.

Si en `50%` aparece `75.0`, significa que la **mediana** es 75.

Además:

- `25%` corresponde al percentil 25.
- `50%` corresponde al percentil 50 y coincide con la mediana.
- `75%` corresponde al percentil 75.

Por lo tanto, `describe()` es una excelente primera herramienta para explorar rápidamente una variable numérica.


### También podemos resumir todas las variables numéricas

Si aplicamos `describe()` directamente al DataFrame, Pandas entrega el resumen de todas las columnas numéricas:


````python
df.describe()
````

## 4. Media aritmética


````python
media = df['Puntaje'].mean()
print("Media del puntaje:", media)
````

### ✏️ ¿Cómo se interpreta?

Si la media fuese 74 puntos, podríamos escribir:

**“El puntaje promedio de los estudiantes es de 74 puntos.”**

La media utiliza todos los valores y puede verse afectada por valores extremadamente grandes o pequeños.


## 5. Mediana


````python
mediana = df['Puntaje'].median()
print("Mediana del puntaje:", mediana)
````

## 6. Moda


````python
moda = df['Puntaje'].mode()
print("Moda:")
print(moda)
````

> `mode()` puede entregar más de un resultado si existen varios valores con la misma frecuencia máxima.


## 7. Comparar media, mediana y moda


````python
print("Media   :", df['Puntaje'].mean())
print("Mediana :", df['Puntaje'].median())
print("Moda    :", df['Puntaje'].mode().tolist())
````

## 8. Efecto de un valor extremo


````python
datos_normales = pd.Series([10, 11, 12, 13, 14])
datos_extremo = pd.Series([10, 11, 12, 13, 100])

print("Sin valor extremo")
print("Media:", datos_normales.mean())
print("Mediana:", datos_normales.median())

print("\nCon valor extremo")
print("Media:", datos_extremo.mean())
print("Mediana:", datos_extremo.median())
````

### ✏️ Idea clave

Un valor extremo puede modificar bastante la **media**, mientras que la **mediana** suele ser más resistente a este tipo de observaciones.


## 9. Percentiles y `quantile()`


````python
p25 = df['Puntaje'].quantile(0.25)
p50 = df['Puntaje'].quantile(0.50)
p75 = df['Puntaje'].quantile(0.75)
p90 = df['Puntaje'].quantile(0.90)

print("P25:", p25)
print("P50:", p50)
print("P75:", p75)
print("P90:", p90)
````

### ✏️ Interpretación de un percentil

Si \(P_{25}=67\), podemos decir aproximadamente:

**“El 25% de los estudiantes tiene un puntaje menor o igual a 67 puntos.”**

### ⚠️ Atención con frases como “10% más alto”

Si queremos separar aproximadamente al **10% con valores más altos**, necesitamos buscar el punto que deja al 90% por debajo:

```python
df['Puntaje'].quantile(0.90)
```

En general, para el **r% superior**, se utiliza el percentil \(100-r\).


## 10. Mínimo y máximo


````python
print("Mínimo:", df['Puntaje'].min())
print("Máximo:", df['Puntaje'].max())
````

Estas funciones son útiles cuando queremos describir el rango de valores observado o complementar la interpretación de un percentil.


`describe()` entrega rápidamente cantidad de datos, media, desviación estándar, mínimo, cuartiles y máximo.

Recuerda:

- `25%` corresponde a \(P_{25}\).
- `50%` corresponde a la mediana.
- `75%` corresponde a \(P_{75}\).


## 12. Filtrar datos


````python
informatica = df[df['Carrera'] == 'Informática']
informatica
````

También podemos combinar condiciones:


````python
seleccion = df[(df['Carrera'] == 'Informática') & (df['Puntaje'] >= 70)]
seleccion
````

## 13. Agrupar datos con `groupby()`


````python
df.groupby('Carrera')['Puntaje'].mean()
````

````python
df.groupby('Carrera')['Puntaje'].median()
````

También podemos calcular un percentil para cada grupo:


````python
df.groupby('Carrera')['Puntaje'].quantile(0.75)
````

## 14. Varias estadísticas por grupo


````python
resumen = df.groupby('Carrera')['Puntaje'].agg(['count', 'mean', 'median', 'min', 'max'])
resumen
````

## 15. Construir un DataFrame resumen


````python
tabla_resumen = pd.DataFrame({
    'Medida': ['Media', 'Mediana', 'P25', 'P75', 'Mínimo', 'Máximo'],
    'Valor': [
        df['Puntaje'].mean(),
        df['Puntaje'].median(),
        df['Puntaje'].quantile(0.25),
        df['Puntaje'].quantile(0.75),
        df['Puntaje'].min(),
        df['Puntaje'].max()
    ]
})

tabla_resumen
````

## 16. Tablas bivariadas con `pd.crosstab()`


````python
pd.crosstab(df['Carrera'], df['Jornada'])
````

También podemos resumir una variable numérica dentro de cada combinación de categorías:


````python
pd.crosstab(
    df['Carrera'],
    df['Jornada'],
    values=df['Puntaje'],
    aggfunc='mean'
)
````

En este caso, cada celda representa el **puntaje promedio** para la combinación correspondiente de carrera y jornada.
