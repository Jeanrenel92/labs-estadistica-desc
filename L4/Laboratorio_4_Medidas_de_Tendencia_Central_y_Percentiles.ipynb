```python
# 4
# media
print(df1["Nota Examen"].mean().round(1))
```

```python
# moda
df1["Nota Examen"].mode()
```

Media: El promedio de las notas del examen el año 2019 fue de 5,1 en la asignatura de álgebra, jornada diurna.

Mediana: El 50% de las notas del examen el año 2019 fue de a lo más 5,2 en la asignatura de álgebra, jornada diurna.

Moda: La nota más frecuente en el examen de álgebra del año 2019, jornada diurna, fue de 5,3.

```python
df1["Prueba 1"].min()
```

```python
#df1["Prueba 1"].quantile(0.6)
print(df1["Prueba 1"].quantile(60/100))
```
