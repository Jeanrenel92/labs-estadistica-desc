## 1. Conceptos Fundamentales de Estadística

-   **Población:** Conjunto total y completo de individuos, elementos,
    transacciones o sucesos que comparten una o más características
    observables sobre los cuales se desea investigar u obtener
    conclusiones.
    -   *Ejemplo:* Todos los clientes suscritos a una empresa de
        telecomunicaciones a nivel nacional.
-   **Muestra:** Subconjunto representativo y medible seleccionado a
    partir de la población objetivo para recolectar datos y llevar a
    cabo el análisis estadístico.
    -   *Ejemplo:* Un grupo de $23.000$ usuarios encuestados o los
        registros contenidos en la base de datos descargada.

## 2. Clasificación de Variables Estadísticas

  -------------------------------------------------------------------------------------------
  Tipo de Variable        Definición y Criterio   Ejemplos en los Laboratorios
  ----------------------- ----------------------- -------------------------------------------
  **Cualitativa Nominal** Atributos, categorías o `Género`, `Carrera`,
                          etiquetas sin orden ni  `Red social preferida`,
                          jerarquía natural.      `Tipo de servicio de Internet` (Fibra
                                                  óptica, DSL).

  **Cualitativa Ordinal** Categorías que poseen   `Nivel de gravedad` (Bajo, Medio, Alto,
                          un orden, jerarquía o   Crítico), `Nivel de satisfacción` (Malo,
                          escala lógica           Regular, Bueno).
                          intrínseca.             

  **Cuantitativa          Valores numéricos que   `Número de materias inscritas`,
  Discreta**              surgen de un proceso de `Cantidad de fallas técnicas reportadas`,
                          **conteo** (números     `Número de hijos`.
                          enteros no negativos).  

  **Cuantitativa          Valores numéricos que   `Edad`, `Horas al día en redes sociales`,
  Continua**              surgen de un proceso de `Cargos mensuales ($)`,
                          **medición** (pueden    `Tiempo de resolución` (horas/minutos).
                          tomar decimales /       
                          infinitos valores       
                          intermedios).           
  -------------------------------------------------------------------------------------------

> **Importancia de la clasificación:** Determina qué cálculos son
> matemáticamente válidos (por ejemplo, no tiene sentido calcular el
> promedio de una variable cualitativa nominal), qué gráficos utilizar
> (gráficos de barras vs. histogramas) y qué modelos analíticos aplicar.

## 3. Medidas Estadísticas y Gráficos según el Tipo de Variable

La elección de una medida estadística o de un gráfico depende del **tipo
de variable**. No todas las operaciones tienen sentido para todas las
variables.

  -----------------------------------------------------------------------
  Tipo de variable        Medidas estadísticas    Gráficos recomendados
                          recomendadas            
  ----------------------- ----------------------- -----------------------
  **Cualitativa Nominal** Frecuencia absoluta,    **Gráfico de barras**,
                          frecuencia relativa,    gráfico de sectores
                          porcentaje, **moda**    

  **Cualitativa Ordinal** Frecuencias,            **Gráfico de barras
                          porcentajes, **moda**,  ordenadas**, gráfico de
                          **mediana** y           sectores
                          percentiles cuando las  
                          categorías tienen un    
                          orden explícitamente    
                          definido                

  **Cuantitativa          **Media**, mediana,     **Gráfico de barras**;
  Discreta**              moda, mínimo, máximo,   histograma cuando
                          rango, varianza,        existen muchos valores
                          desviación estándar,    posibles; boxplot
                          cuartiles y percentiles

  **Cuantitativa          **Media**, mediana,     **Histograma**,
  Continua**              moda, mínimo, máximo,   boxplot, gráfico de
                          rango, varianza,        densidad
                          desviación estándar,    
                          cuartiles y percentiles
  -----------------------------------------------------------------------

> **Regla práctica:** para variables cualitativas interesa
> principalmente **contar categorías**. Para variables cuantitativas
> interesa además estudiar el **centro**, la **dispersión** y la **forma
> de la distribución**.

### Formulario de medidas estadísticas en Pandas

``` python
# Frecuencia absoluta
df['variable'].value_counts()

# Frecuencia relativa
df['variable'].value_counts(normalize=True)

# Porcentaje
df['variable'].value_counts(normalize=True) * 100

# Moda
df['variable'].mode()

# Media
df['variable'].mean()

# Mediana
df['variable'].median()

# Mínimo y máximo
df['variable'].min()
df['variable'].max()

# Rango
df['variable'].max() - df['variable'].min()

# Varianza
df['variable'].var()

# Desviación estándar
df['variable'].std()

# Cuartiles
df['variable'].quantile([0.25, 0.50, 0.75])

# Resumen descriptivo
df['variable'].describe()
```

### Gráficos básicos con Pandas y Matplotlib

``` python
import matplotlib.pyplot as plt

# Cualitativa: gráfico de barras
df['Carrera'].value_counts().plot(kind='bar')
plt.xlabel('Carrera')
plt.ylabel('Frecuencia')
plt.title('Frecuencia por carrera')
plt.show()

# Cuantitativa: histograma
df['Edad'].plot(kind='hist', bins=10)
plt.xlabel('Edad')
plt.title('Distribución de la edad')
plt.show()

# Cuantitativa: boxplot
df['Edad'].plot(kind='box')
plt.title('Boxplot de la edad')
plt.show()
```

## 4. Formulario Esencial de Pandas (Python)

### A. Carga, Creación e Inspección Estructural

``` python
import pandas as pd

# 1. Crear un DataFrame desde un diccionario
df = pd.DataFrame({
    'Nombre': ['Rodrigo', 'Pedro', 'Ignacia'],
    'Edad': [21, 23, 20],
    'Carrera': ['Informática', 'Telecomunicaciones', 'Informática']
})

# 2. Cargar archivo Excel
df = pd.read_excel('clientes.xlsx')

# 3. Exploración de filas y dimensiones
df.head(5)          # Primeras 5 filas
df.tail(5)          # Últimas 5 filas
df.shape            # Tupla: (número_filas, número_columnas)
type(df)            # Tipo de objeto (pandas.core.frame.DataFrame)

# 4. Estructura y tipos de datos
df.info()           # Resumen de columnas, tipos y valores no nulos
df.dtypes           # Tipo de dato exacto de cada columna
```

------------------------------------------------------------------------

### B. Selección, Conteo y Estadísticas Básicas

``` python
# Seleccionar una columna (Serie) o múltiples columnas (DataFrame)
df['Carrera']
df[['Nombre', 'Carrera', 'Edad']]

# Conteo de frecuencias por categoría
df['Carrera'].value_counts()

# Valor más frecuente (Moda)
df['Edad'].mode()[0]

# Cálculo de promedio (media aritmética)
promedio_cargos = df['Cargos mensuales'].mean()
```

------------------------------------------------------------------------

### C. Modificación Permanente vs. Temporal

En Pandas, la mayoría de los métodos devuelven una **copia** modificada
y no alteran el DataFrame original a menos que se aplique una de las
siguientes técnicas:

1.  **Reasignación (Método recomendado):**

    ``` python
    # Cambio de tipo de dato (Casting)
    df['Antigüedad'] = df['Antigüedad'].astype(float)

    # Crear una nueva columna calculada
    df['Cargos acumulados'] = df['Cargos mensuales'] * df['Antigüedad']

    # Sobreescribir el DataFrame con un filtro permanente
    df = df[df['Servicio de Internet'] == 'Fibra óptica']
    ```

2.  **Parámetro `inplace=True` (cuando está disponible):**

    ``` python
    # Eliminar columnas de forma directa
    df.drop(columns=['ColumnaInnecesaria'], inplace=True)

    # Renombrar columnas
    df.rename(columns={'Edad': 'Edad_Años'}, inplace=True)
    ```

------------------------------------------------------------------------

### D. Filtrado Condicional (Simple y Múltiple)

``` python
# Filtrado simple (una condición)
filtro_fibra = df[df['Servicio de Internet'] == 'Fibra óptica']

# Filtrado múltiple (usar paréntesis individuales obligatoriamente)
# Operador &: AND (todas las condiciones se deben cumplir)
# Operador |: OR  (al menos una condición se debe cumplir)
condicion = (
    (df['Servicio de Internet'] == 'Fibra óptica') &
    (df['Protección Malware'] == 'Sí') &
    (df['Deserción Cliente'] == 'No')
)

df_segmentado = df[condicion]
promedio_segmento = df_segmentado['Cargos mensuales'].mean()

print(f"Total registros: {len(df_segmentado)}")
print(f"Promedio de cobro mensual: ${promedio_segmento:.2f}")
```

------------------------------------------------------------------------

## 5. Regla de Oro para el Análisis de Datos

``` text
Tipo Estadístico          Tipo en Pandas          Operaciones Válidas
-----------------------------------------------------------------------------
Cualitativa Nominal   ->  object / string    ->   value_counts(), mode()
Cualitativa Ordinal   ->  category ordenada  ->   value_counts(), mode(); median() si el orden está definido
Cuantitativa          ->  int64 / float64    ->   mean(), median(), var(), std(), quantile(), describe()
```
