## 3. Media

La **media aritmética** corresponde al promedio de los datos:

$$
\bar{x}=\frac{x_1+x_2+\cdots+x_n}{n}
       =\frac{1}{n}\sum_{i=1}^{n}x_i.
$$

En `pandas`:

```python
df["Pago base"].mean()
```

### Interpretación

La media representa un valor central de los datos, pero puede verse afectada por valores extremadamente altos o bajos.

Si la variable corresponde a dinero, el resultado debe expresarse con **\$**.

Por ejemplo:

$$
\bar{x}=\$75\,000.
$$

````python
media = df["Pago base"].mean()
print(f"Media: ${media:,.2f}")
````

## 4. Mediana

La **mediana** es el valor que ocupa la posición central después de ordenar los datos.

En `pandas`:

```python
df["Pago base"].median()
```

La mediana es menos sensible a valores extremos que la media.

Por ejemplo:

$$
\operatorname{Mediana}=\$68\,500.
$$

### Pregunta útil al interpretar

Si la media y la mediana son muy diferentes, ¿podrían existir valores extremos o una distribución asimétrica?

````python
mediana = df["Pago base"].median()
print(f"Mediana: ${mediana:,.2f}")
````

## 5. Rango

El **rango** mide la distancia entre el valor máximo y el valor mínimo:

$$
R=x_{\max}-x_{\min}.
$$

En `pandas`:

```python
df["Pago base"].max() - df["Pago base"].min()
```

Como el rango conserva la unidad de la variable, si estamos trabajando con dinero debe escribirse con **\$**.

Por ejemplo:

$$
R=\$120\,000.
$$

### Interpretación

Un rango mayor indica una mayor separación entre los valores extremos.

**Precaución:** el rango utiliza solamente el máximo y el mínimo; no describe por sí solo cómo se distribuyen todos los demás datos.

````python
rango = df["Pago base"].max() - df["Pago base"].min()
print(f"Rango: ${rango:,.2f}")
````

## 6. Desviación estándar

La **desviación estándar muestral** mide la dispersión de los datos alrededor de la media:

$$
s=
\sqrt{
\frac{\displaystyle\sum_{i=1}^{n}(x_i-\bar{x})^2}
{n-1}
}.
$$

En `pandas`:

```python
df["Pago base"].std()
```

### Interpretación

- Una desviación estándar pequeña indica que los datos tienden a estar más próximos a la media.
- Una desviación estándar grande indica mayor dispersión.

La desviación estándar mantiene la unidad de la variable original. Para dinero:

$$
s=\$25\,400.
$$

````python
desviacion = df["Pago base"].std()
print(f"Desviación estándar: ${desviacion:,.2f}")
````

## 7. Coeficiente de variación

El **coeficiente de variación** compara la desviación estándar con la media:

$$
CV=\frac{s}{\bar{x}}\cdot100\%.
$$

En Python:

```python
cv = df["Pago base"].std() / df["Pago base"].mean() * 100
```

### Interpretación

$$
\boxed{\text{Mayor }CV \Longrightarrow \text{mayor dispersión relativa}}
$$

$$
\boxed{\text{Menor }CV \Longrightarrow \text{mayor homogeneidad}}
$$

A diferencia de la media, mediana, rango y desviación estándar, el coeficiente de variación **no lleva signo \$**.

Se expresa como porcentaje:

$$
CV=32.5\%.
$$

````python
cv = df["Pago base"].std() / df["Pago base"].mean() * 100
print(f"Coeficiente de variación: {cv:.2f}%")
````

## 9. Filtrar datos

Una vez identificadas las categorías que se quieren estudiar, podemos seleccionar solamente esas observaciones.

Por ejemplo:

```python
cargos = df["Título profesional"].value_counts().head(4).index
df_filtrado = df[df["Título profesional"].isin(cargos)]
```

El método `.isin()` pregunta si cada valor pertenece al conjunto indicado.

````python
cargos = df["Título profesional"].value_counts().head(4).index
df_filtrado = df[df["Título profesional"].isin(cargos)]

df_filtrado.head()
````

## 10. Agrupar datos

`groupby()` permite separar los datos en grupos y calcular estadísticas para cada uno.

Por ejemplo, para obtener media y desviación estándar según una variable:

```python
df.groupby("Año")["Pago base"].agg(["mean", "std"])
```

La estructura general es:

```python
df.groupby("variable_para_agrupar")["variable_numerica"].agg(...)
```

Esto resulta útil cuando se necesita **comparar grupos**.

````python
tabla = df.groupby("Año")["Pago base"].agg(["mean", "std"])
tabla
````

## 11. Calcular el CV para cada grupo

Después de calcular la media y la desviación estándar de cada grupo:

```python
tabla = df.groupby("Año")["Pago base"].agg(["mean", "std"])
```

podemos agregar el coeficiente de variación:

```python
tabla["CV"] = tabla["std"] / tabla["mean"] * 100
```

Luego se comparan los valores obtenidos.

Recuerda:

$$
CV=\frac{s}{\bar{x}}\cdot100\%.
$$

````python
tabla = df.groupby("Año")["Pago base"].agg(["mean", "std"])
tabla["CV"] = tabla["std"] / tabla["mean"] * 100

tabla
````

## 12. Comparar categorías

La misma herramienta puede utilizarse con una variable categórica, por ejemplo:

```python
tabla = df_filtrado.groupby("Título profesional")["Pago base"].agg(
    ["mean", "median", "std"]
)
```

Luego:

```python
tabla["CV"] = tabla["std"] / tabla["mean"] * 100
```

Con esta tabla puedes comparar:

- media;
- mediana;
- desviación estándar;
- coeficiente de variación.

No basta con mostrar los números: debes **interpretar qué indican**.

````python
tabla = df_filtrado.groupby("Título profesional")["Pago base"].agg(
    ["mean", "median", "std"]
)

tabla["CV"] = tabla["std"] / tabla["mean"] * 100
tabla
````

## 13. ¿Qué medida utilizar?

| Medida | ¿Qué informa? | Unidad |
|---|---|---|
| Media | Centro o promedio | \$ |
| Mediana | Valor central | \$ |
| Rango | Separación entre máximo y mínimo | \$ |
| Desviación estándar | Dispersión absoluta respecto de la media | \$ |
| Coeficiente de variación | Dispersión relativa | % |

### Ideas clave

$$
\boxed{\text{Mayor desviación estándar}
\Longrightarrow
\text{mayor dispersión absoluta}}
$$

$$
\boxed{\text{Mayor CV}
\Longrightarrow
\text{mayor dispersión relativa}}
$$

$$
\boxed{\text{Menor CV}
\Longrightarrow
\text{mayor homogeneidad}}
$$
